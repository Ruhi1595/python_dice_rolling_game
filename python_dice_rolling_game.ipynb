#  Python Dice Rolling Simulator

An interactive command-line Python game that simulates rolling two standard six-sided dice (`1–6`).

### Features
- **Continuous Game Loop:** Prompts the user repeatedly (`y/n`) until they choose to exit.
- **Input Normalization & Validation:** Handles case-insensitivity, accidental whitespace, and invalid entries gracefully.
- **Random Simulation:** Uses Python's built-in `random` module to generate independent dice outcomes.

```python
import random

# Continuously prompt the user until they choose to exit
while True:
    # Normalize input to lowercase and remove surrounding whitespace
    choice = input('Roll the dice? (y/n): ').strip().lower()

    if choice == 'y':
        # Simulate rolling two standard six-sided dice
        die1 = random.randint(1, 6)
        die2 = random.randint(1, 6)
        print(f'({die1}, {die2})')
    elif choice == 'n':
        # Exit the game loop cleanly
        print('Thanks for playing')
        break
    else:
        # Handle any unexpected user input
        print('Invalid choice')
```
